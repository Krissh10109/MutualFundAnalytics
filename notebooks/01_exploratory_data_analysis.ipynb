# Mutual Fund Analytics - Exploratory Data Analysis (EDA)

This notebook conducts exploratory analysis on mutual fund NAV time series, category returns, and risk metrics.

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add root directory to sys.path
sys.path.append(os.path.abspath('..'))
from scripts.data_loader import generate_synthetic_fund_data, preprocess_nav_data
from scripts.metrics_calculator import evaluate_fund_performance

sns.set_theme(style='darkgrid')
```

## 1. Load Data

```python
df_funds, df_bench = generate_synthetic_fund_data(days=730)
pivot_nav = preprocess_nav_data(df_funds)
pivot_nav.head()
```

## 2. Visualize Historical NAV Growth

```python
plt.figure(figsize=(12, 6))
for column in pivot_nav.columns:
    plt.plot(pivot_nav.index, pivot_nav[column], label=column)

plt.title('Mutual Fund NAV Trajectory Over Time', fontsize=14, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('NAV (INR)')
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Daily Returns Distribution

```python
daily_returns = pivot_nav.pct_change().dropna()
plt.figure(figsize=(10, 5))
sns.boxplot(data=daily_returns * 100)
plt.title('Daily Return Distribution by Fund (%)')
plt.ylabel('Daily Return (%)')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```
